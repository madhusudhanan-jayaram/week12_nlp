# Fake News Detection Project
This notebook demonstrates a workflow for detecting fake news using machine learning techniques. Each step is explained with professional comments for clarity.

```python
import pandas as pd
import numpy as np
import itertools
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn import metrics
import matplotlib.pyplot as plt
```

## Import Required Libraries
Import essential libraries for data manipulation, machine learning, and visualization.

```python
df=pd.read_csv('news.csv', index_col=None)
```

## Load the Dataset
Read the news dataset from a CSV file into a pandas DataFrame for further analysis.

```python
df
```

## Display the DataFrame
Show the loaded DataFrame to inspect the data and verify successful loading.

```python
dataset = df.drop("Unnamed: 0", axis=1)  # Remove the unnecessary 'Unnamed: 0' column from the DataFrame (axis=1 specifies column-wise operation)
```

## Data Cleaning
Remove unnecessary columns from the DataFrame to prepare the dataset for modeling.

```python
dataset
```

## Display Cleaned Dataset
Show the cleaned dataset to confirm removal of unnecessary columns.

```python
y=dataset["label"]
```

## Extract Target Variable
Separate the target variable (`label`) from the dataset for supervised learning.

```python
X_train, X_test, y_train, y_test = train_test_split(dataset['text'], y, test_size=0.33, random_state=53)
```

## Split Data into Training and Test Sets
Divide the dataset into training and test sets to evaluate model performance objectively.

```python
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf_train = tfidf_vectorizer.fit_transform(X_train)
tfidf_test = tfidf_vectorizer.transform(X_test)
```

## Text Vectorization
Convert text data into numerical features using CountVectorizer for machine learning algorithms.

```python
dir(tfidf_train)
```

## Inspect Vectorized Data
Explore the properties and structure of the vectorized training data.

```python
len(tfidf_vectorizer.get_feature_names_out())
```

## Feature Analysis
Determine the number of unique features (words) extracted from the text data.

```python
print(tfidf_train.toarray())
```

## View Vectorized Data Array
Display the array representation of the vectorized training data for inspection.

```python
clf = MultinomialNB()

clf.fit(tfidf_train, y_train)
pred = clf.predict(tfidf_test)
score = metrics.accuracy_score(y_test, pred)
print("accuracy:   %0.3f" % score)

cm = metrics.confusion_matrix(y_test, pred, labels=['FAKE', 'REAL'])
```

## Train and Evaluate Model
Train a Multinomial Naive Bayes classifier and evaluate its accuracy using the test set. Display the confusion matrix for further analysis.

```python
from sklearn.metrics import classification_report

report=classification_report(y_test, pred)
```

Produce a detailed classification report to assess model performance on various metrics.
